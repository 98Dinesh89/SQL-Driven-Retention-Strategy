# Handel Missing values in 'Review Rating'
using Median Imputation

```python
df_clean["Review Rating"].isnull().sum()
```

```python
median_rating = df_clean["Review Rating"].median()
print(median_rating)
```

```python
df_clean["Review Rating"] = df_clean["Review Rating"].fillna(median_rating)
df_clean["Review Rating"].isnull().sum()
```

# Standradize Text Columns

```python
cat_cols = [
    "Gender",
    "Item Purchased",
    "Category",
    "Location",
    "Size",
    "Color",
    "Season",
    "Subscription Status",
    "Shipping Type",
    "Discount Applied",
    "Promo Code Used",
    "Payment Method",
    "Frequency of Purchases"
]

for col in cat_cols:
    df_clean[col] = (
        df_clean[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )
```

# Outlier Detection
using IQR based outlire detection

```python
def detect_outliers_iqr(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
```
